```python
from langgraph.store.memory import InMemoryStore
```

```python
# create a store object
store = InMemoryStore()
```

```python
# create a namespace.
namespace = ('user', 'u1')
```

### Creating Memories

```python
# add the memories
store.put(namespace, "1",{"data": "user likes pizzz"})
store.put(namespace, "2",{"data": "user prefered dark mode."})
```

```python
# another namespace
namespace2 = ("user", "u2")

# Add memories
store.put(namespace2, "1", {"data": "User likes pasta"})
store.put(namespace2, "2", {"data": "User prefers grid style navigation"})
```

### Retrieving Memories

```python
store.get(namespace, "1")
```

```python
# store.get(namespace, key)
store.get(namespace2, "2")
```

### Retrieving all Memories

```python
items = store.search(namespace2)

for item in items:
    print(item.value)
```
