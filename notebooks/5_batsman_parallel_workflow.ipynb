```python
from langgraph.graph import  StateGraph, START, END
from typing import TypedDict
```

```python
# define the state

class BatsmanState(TypedDict):

    runs: int
    balls: int
    fours: int
    sixes: int

    sr: float
    bpb: float
    boundary_pect: float
```

```python
def calculate_sr(state: BatsmanState):

    sr = (state['runs']/state['balls'])/100
    state['sr'] = sr
    return {'sr': sr}
    

def calculate_brp(state: BatsmanState):
    bpb = (state['balls'])/(state['fours'] + state['sixes'])
    state['bpb'] = bpb
    return {'bpb':bpb}


def calculate_boundary_pect(state: BatsmanState):
    boundry_percentage = ((state['fours']*4) + (state['sixes']* 6)/state['runs'])*100
    state['boundary_pect'] = boundry_percentage
    return {'boundary_pect': boundry_percentage
            }

def summary(state: BatsmanState):

    summary = f"""
Strike Rate - {state['sr']} \n
Balls per boundary - {state['bpb']} \n
Boundary percent - {state['boundary_pect']}
"""
    
    return {'summary': summary}

```

```python
# define the stategraph:


graph = StateGraph(BatsmanState)

# add node
graph.add_node('calculate_sr', calculate_sr)
graph.add_node('calculate_brp',calculate_brp)
graph.add_node('calculate_boundary_pect',calculate_boundary_pect )
graph.add_node('summary', summary)

# add edge
graph.add_edge(START, 'calculate_sr')
graph.add_edge(START, 'calculate_brp')
graph.add_edge(START, 'calculate_boundary_pect')

graph.add_edge('calculate_sr', 'summary')
graph.add_edge('calculate_brp','summary')
graph.add_edge('calculate_boundary_pect','summary')

graph.add_edge('summary', END)


workflow = graph.compile()
```

```python
intial_state = {
    'runs': 100,
    'balls': 50,
    'fours': 6,
    'sixes': 4
}

workflow.invoke(intial_state)
```
